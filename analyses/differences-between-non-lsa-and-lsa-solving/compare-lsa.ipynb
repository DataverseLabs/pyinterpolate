```python
import pandas as pd
```

```python
f1 = 'lsa-comp-dem2180.csv'
f2 = 'lsa-comp-meuse.csv'
```

```python
df1 = pd.read_csv(f1, index_col=False)
df1.drop(columns=['Unnamed: 0'], inplace=True)
df1.head()
```

```python
df2 = pd.read_csv(f2, index_col=False)
df2.drop(columns=['Unnamed: 0'], inplace=True)
df2.head()
```

```python
df = pd.concat([df1, df2], axis=0)
```

```python
len(df) == (len(df1) + len(df2))
```

```python
df.head()
```

```python
# average RMSE per nn
```

```python
avg_rmse_nn = df.groupby('nn')[['rmse', 'rmse_lsa']].mean()
```

```python
avg_rmse_nn['rmse'] - avg_rmse_nn['rmse_lsa']
```

```python
# average time per nn
```

```python
avg_time_nn = df.groupby('nn')[['time-linalgsolve', 'time-lsa']].mean()
```

```python
avg_time_nn['time-linalgsolve'] - avg_time_nn['time-lsa']
```
